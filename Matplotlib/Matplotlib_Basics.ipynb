# **Matplotlib Basics**

Matplotlib is a Python library for creating charts and plots.

This notebook contains simple examples and notes to practice data visualization using `matplotlib.pyplot`.

## **Basic Line Plot**

This example shows daily study hours using a line plot. It demonstrates how to add a title, axis labels, a legend, and customize axis ticks and line appearance.

```python
import matplotlib.pyplot as plt
```

```python
days = ["Sat", "Sun", "Mon", "Tue", "Wed"]
study_hours = [2, 3, 1, 4, 3]

# Create a figure and set its width and height in inches.
plt.figure(figsize=(8, 4))

# Plot study hours and customize the line and data-point markers.
plt.plot(
    days,
    study_hours,
    label="My Study Hours",
    color="purple",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=7
)

# Add a title and control its font size and spacing above the plot.
plt.title("Study Hours During the Week", fontsize=14, pad=15)

# Label the horizontal and vertical axes.
plt.xlabel("Day", fontsize=12)
plt.ylabel("Study Hours", fontsize=12)

# Display the line label in a legend at the lower-right corner.
plt.legend(loc="lower right", fontsize=10)

# Show all day labels and set their rotation and font size.
plt.xticks(days, rotation=0, fontsize=10)

# Choose which values appear as ticks on the vertical axis.
plt.yticks([0, 1, 2, 3, 4, 5])

# Set the visible range of the vertical axis.
plt.ylim(0, 5)

# Add horizontal grid lines to make values easier to read.
plt.grid(axis="y", linestyle=":", alpha=0.5)

# Adjust spacing to help prevent labels from being clipped.
plt.tight_layout()

# Display the figure.
plt.show()
```

## **Two Lines on the Same Plot**

Use `plt.plot()` twice to compare Math and Science grades on the same graph. Add a label for each line and display them using `plt.legend()`.

```python
from matplotlib.lines import lineStyles
students = ["Sara", "Omar", "Mona", "Ali", "Nour"]
math_grades = [85, 70, 95, 60, 80]
science_grades = [75, 85, 90, 70, 85]
plt.plot(students,math_grades,'r*--', label='Math')
# 'r*--' :Red line (r), star markers (*), dashed style (--)
plt.plot(students,science_grades,'g.:',label='Science' )

plt.title('Math and Science Grades',
          fontdict={'fontweight' : 'bold' , 'fontsize':15, 'color':'black'})
plt.xlabel('Student',fontsize=10)
plt.ylabel('Grade',fontsize=10)
plt.legend(loc='lower right')

plt.ylim(0,100)
plt.grid(axis="y", linestyle=":", alpha=0.5)

plt.show()

```

### **Multiple Lines on One Plot**
Plot t, t², and t³ on the same axes using a single `plt.plot()` call.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0., 5., 0.2)
```

```python
# Plot t, t², and t³ on the same graph
plt.plot(t,t,'m*--',t,t**2,'y>',t,t**3,'bs')
plt.show()
```

### **Separate Figures**
Create three separate figures, each displaying a different function: t, t², and t³.

```python
plt.figure()
plt.plot(t, t, 'm*--')
plt.title("Linear")
plt.show()

plt.figure()
plt.plot(t, t**2, 'y>')
plt.title("Quadratic")
plt.show()

plt.figure()
plt.plot(t, t**3, 'bs')
plt.title("cubic")
plt.show()
```

### **Subplots in One Figure**


### Horizontal Subplots
Display three plots side by side in one figure using one row and three columns.

```python
# Create a figure with a width of 12 inches and a height of 4 inches.
plt.figure(figsize=(12, 4))

# Select position 1 in a 1-row, 3-column
plt.subplot(1, 3, 1)
plt.plot(t, t, 'm*--')
plt.title("linear")
plt.xlabel("t values")
plt.ylabel("output values")

# Select position 2
plt.subplot(1, 3, 2)
plt.plot(t, t**2, 'y>')
plt.title("Quadratic")
plt.xlabel("t values")
plt.ylabel("output values")

# Select position 3
plt.subplot(1, 3, 3)
plt.plot(t, t**3, 'bs')
plt.title("Cubic")
plt.xlabel("t values")
plt.ylabel("output values")

# Adjust spacing between the plots.
plt.tight_layout()

# Display the figure with all three subplots.
plt.show()
```

### Vertical Subplots
Display three plots vertically in one figure using three rows and one column.

```python
from matplotlib.figure import SubFigure
plt.figure(figsize=(4,10))

plt.subplot(3,1,1)
plt.plot(t, t, 'm*--')
plt.title("linear")
plt.xlabel("t values")
plt.ylabel("output values")

plt.subplot(3, 1, 2)
plt.plot(t, t**2, 'y>')
plt.title("Quadratic")
plt.xlabel("t values")
plt.ylabel("output values")

plt.subplot(3, 1, 3)
plt.plot(t, t**3, 'bs')
plt.title("Cubic")
plt.xlabel("t values")
plt.ylabel("output values")

plt.tight_layout()

plt.show()
```

## **Figures, Axes, and Subplots**

**A Figure** is the entire drawing area.
**An Axes** is an individual plotting area with its own axes, title, and data.
**Subplots** organize multiple Axes into rows and columns within one Figure.

`plt.subplots()` creates the Figure and its Axes together.

```python
# Generate 100 evenly spaced values from 1 to 5,
# including both endpoints.
x=np.linspace(1,10,100)

fig , axes = plt.subplots(2,3,figsize=(12,8))

axes[0,0].plot(x,np.sin(x))
axes[0,0].set_title("Sin Func")

axes[0,1].plot(x,np.cos(x))
axes[0,1].set_title("Cos Func")

axes[0,2].plot(x,np.tan(x))
axes[0,2].set_title("Tan Func")

axes[1,0].plot(x,np.exp(x))
axes[1,0].set_title("Exp Func")

axes[1,1].plot(x,np.log(x))
axes[1,1].set_title("Log Func")

axes[1,2].plot(x,x)
axes[1,2].set_title("Linear Func")

plt.tight_layout()
plt.show()
```

## **Scatter Plot**
A scatter plot displays data as individual points to show the relationship between two variables.

```python
import numpy as np
x=np.random.rand(100) * 10
y=np.random.rand(100) * 10

plt.figure(figsize=(5, 4))

plt.scatter(x,y,marker='D',color='m',edgecolors='#000000',s=20,linewidths=1.5 , alpha=0.5)
plt.show()
```

## **Bar Chart**
A bar chart uses bars to compare values across different categories.

```python
subjects = ["Math", "Python", "ML", "CV"]
study_hours = [3, 5, 4, 2]

plt.figure(figsize=(5, 4))

# Create bars with different colors and black borders.
bars = plt.bar(
    subjects,
    study_hours,
    width=0.3,
    edgecolor="black",
    linewidth=1.2
)

colors=["purple", "pink", "skyblue", "orange"]
hatches=["/", "\\", "x", "."]

for bar, color,hatch in zip(bars,colors,hatches):
  # set_facecolor():
  # to Change the fill color while keeping the black border.
  bar.set_facecolor(color)
  bar.set_hatch(hatch)


# Add a title and axis labels.
plt.title("Study Hours by Subject")
plt.xlabel("Subject")
plt.ylabel("Study Hours")

# Set the vertical axis ticks.
plt.yticks(range(0, 7))

plt.tight_layout()
plt.show()
```

```python
# Create horizontal bars.
plt.barh(subjects, study_hours, height=0.5, color="purple")
plt.xlabel("Study Hours")
plt.ylabel("Subject")
plt.tight_layout()
plt.show()
```

## **Histogram**
A histogram shows the distribution of numerical data by grouping values into intervals called bins.

```python
grades = [52, 57, 61, 65, 65, 68, 72, 74, 78, 81, 85, 88, 92, 96]

plt.figure(figsize=(7, 4))

# Group grades into intervals and count the values in each interval.
plt.hist(
    grades,
    bins=[50, 60, 70, 80, 90, 100],
    color="skyblue",
    edgecolor="black",
    linewidth=1,
    alpha=0.8
)

plt.title("Distribution of Student Grades")
plt.xlabel("Grade")
plt.ylabel("Number of Students")

plt.xticks([50, 60, 70, 80, 90, 100])
plt.yticks([0,1,2,3,4])

plt.tight_layout()
plt.show()
```

```python
plt.hist(
    grades,
    bins=[50, 60, 70, 80, 90, 100],
    color="skyblue",
    edgecolor="black",
    linewidth=1,
    alpha=0.8,
    orientation='horizontal'
)
plt.show()
```

## **Pie Chart**
A pie chart shows how a total is divided among different categories.

```python
subjects = ["Math", "Python", "ML", "CV"]
study_hours = [3, 5, 4, 2]

plt.figure(figsize=(6, 6))

plt.pie(
    study_hours,
    labels=subjects,
    colors=["purple", "pink", "skyblue", "orange"],
    autopct="%.1f%%",
    startangle=90,
    explode=[0, 0.1, 0, 0],
    wedgeprops={"edgecolor": "black", "linewidth": 1.2}
)

plt.title("Study Time Distribution")

plt.show()
```
