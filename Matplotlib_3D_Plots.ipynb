## **3D Scatter Plot**
A 3D scatter plot displays points using three coordinates: X, Y, and Z, to explore the relationship between three variables.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
study_hours = [1, 2, 3, 4, 5]
sleep_hours = [6, 7, 5, 8, 7]
grades = [50, 60, 55, 80, 85]

# Create the figure.
fig = plt.figure(figsize=(8, 6))

# Create a three-dimensional plotting area.
ax = fig.add_subplot(1, 1, 1, projection="3d")

# Plot the X, Y, and Z values as individual points.
ax.scatter(
    study_hours,
    sleep_hours,
    grades,
    color="purple",
    s=60
)

ax.set_title("Study Hours, Sleep Hours, and Grades")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Sleep Hours")
ax.set_zlabel("Grade")

plt.show()
```

## **3D Curve Plot**
A 3D curve plot connects points in order using X, Y, and Z coordinates. This example creates a helix using sine and cosine for circular motion and increasing Z values for height.

```python
t = np.linspace(0, 4 * np.pi, 200)

x = np.cos(t)
y = np.sin(t)
z = t

# Create the figure and a 3D plotting area.
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(1, 1, 1, projection="3d")

# Connect the points in order to create a 3D curve.
ax.plot(x, y, z, color="purple", linewidth=2, label="Helix")

ax.set_title("3D Helix")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.legend()

plt.legend()
plt.show()
```

## **3D Bar Plot**
A 3D bar plot places bars at X and Y positions, with each bar's height representing a value along the Z-axis.

| Argument | Description |
|----------|-------------|
| `x`, `y` | The starting position of each bar on the XY plane. |
| `z` | The starting height of each bar; zero in this example. |
| `dx` | The width of each bar along the X-axis. |
| `dy` | The depth of each bar along the Y-axis. |
| `dz` | The height of each bar, representing its value. |

```python
x = np.array([0, 1, 0, 1])
y = np.array([0, 0, 1, 1])

z = np.zeros(4)

# Set the width, depth, and height of the bars.
dx = 0.4
dy = 0.4
dz = np.array([2, 4, 3, 5])

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(1, 1, 1, projection="3d")

# Draw the bars using their positions and dimensions.
ax.bar3d(
    x, y, z,
    dx, dy, dz,
    color=["purple", "skyblue", "purple", "skyblue"],
    edgecolor="black",
)

# Place subject labels at the centers of the bars.
ax.set_xticks([0.2, 1.2], labels=["Math", "Python"])

# Place day labels at the centers of the bars.
ax.set_yticks([0.2, 1.2], labels=["Saturday", "Sunday"])

ax.set_title("Study Hours by Subject and Day")
ax.set_xlabel("Subject",color='r')
ax.set_ylabel("Day",color='r')
ax.set_zlabel("Study Hours",color='r')



plt.show()
```

## **3D Surface Plot**
A 3D surface plot displays Z values as heights over a grid of X and Y coordinates. Colors can help show differences in height.

```python
# Generate evenly spaced values along both axes.
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

# Create a grid containing all combinations of X and Y positions.
X, Y = np.meshgrid(x, y)

# Calculate the height at each grid position.
Z = X**2 + Y**2

# Create a figure and a 3D plotting area.
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(1, 1, 1, projection="3d")

# Draw the surface and color it according to height.
surface = ax.plot_surface(
    X, Y, Z,
    cmap="viridis",
    edgecolor="none",
    alpha=0.9
)

# Add a colorbar showing the values represented by the colors.
fig.colorbar(surface, ax=ax, label="Height (Z)", shrink=0.6)

ax.set_title("3D Surface Plot")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.tight_layout()
plt.show()
```
